```python
# Como sumamos a una lista un valor escalar?
arr1 = [x for x in range(4)]
## arr2 = arr1 + 8
print("arr1 :", arr1)
```

```python
# SUMA CON NUMPY
np_arr1 = np.array(arr1)
print("np_arr1 :", np_arr1)

# Con ndarray de NumPy, podemos sumar un valor escalar a todos los elementos del array
np_arr2 = np_arr1 + 8
print("np_arr2 :", np_arr2)
```

```python
# RESTA CON NUMPY
np_arr1 = np.array(arr1)
print("np_arr1 :", np_arr1)

# Con ndarray de NumPy, podemos restar un valor escalar a todos los elementos del array
np_arr2 = np_arr1 - 8
print("np_arr2 :", np_arr2)
```

```python
# MULTIPLICACIÓN CON NUMPY
np_arr1 = np.array(arr1)
print("np_arr1 :", np_arr1)

# Con ndarray de NumPy, podemos multiplicar todos los elementos del array por un valor escalar
np_arr2 = np_arr1 * 8
print("np_arr2 :", np_arr2)
```

```python
# DIVISIÓN CON NUMPY
np_arr1 = np.array(arr1)
print("np_arr1 :", np_arr1)

# Con ndarray de NumPy, podemos dividir todos los elementos del array por un valor escalar
np_arr2 = np_arr1 / 8
print("np_arr2 :", np_arr2)
```

```python
np_arr1 = np.arange(4)
np_arr2 = np.arange(4, 8)

print("np_arr1 :", np_arr1)
print("np_arr2 :", np_arr2)
```

```python
# Producto punto de np_arr1 y np_arr2
print(sum(np_arr1 * np_arr2))

# Producto punto de np_arr1 y np_arr2
print(np.dot(np_arr1, np_arr2))
print(np_arr1.dot(np_arr2))

# Producto punto de np_arr2 y np_arr1
print(np.dot(np_arr2, np_arr1))
print(np_arr2.dot(np_arr1))
```
