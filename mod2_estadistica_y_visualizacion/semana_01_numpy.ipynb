**7. Iteraciones por columnas y filas**

Objetivo: practicar agregaciones por ejes (axis=0 y axis=1) en matrices.

```python
a_arr = np.arange(0,12).reshape((3,4))
print("a_arr :\n", a_arr)
```

```python
# apply_along_axis : Función que aplica una función a lo largo de un eje de un arreglo

# Sumas

COL = 0 # Para procesar por columnas, el eje es 0
FILA = 1 # Para procesar por filas, el eje es 1

# Procesar por columnas (eje 0), para calcular la SUMA de cada columna
suma_columnas = np.apply_along_axis(np.sum, axis=COL, arr=a_arr)
print("Suma de cada columna:", suma_columnas)

# Procesar por filas (eje 1), para calcular la SUMA de cada fila
suma_filas = np.apply_along_axis(np.sum, axis=FILA, arr=a_arr)
print("Suma de cada fila:", suma_filas)
```

```python
# Promedios

# Procesar por columnas (eje 0), para calcular el PROMEDIO de cada columna
promedio_columnas = np.apply_along_axis(np.mean, axis=COL, arr=a_arr)
print("Promedio de cada columna:", promedio_columnas)

# Procesar por filas (eje 1), para calcular el PROMEDIO de cada fila
promedio_filas = np.apply_along_axis(np.mean, axis=FILA, arr=a_arr)
print("Promedio de cada fila:", promedio_filas)
```

```python
def mi_funcion(x):
    #print("x :", x)
    return x ** 2

# Procesar por columnas (eje 0), para calcular el cuadrado de cada elemento de cada columna
cuadrado_columnas = np.apply_along_axis(mi_funcion, axis=COL, arr=a_arr)
print("Cuadrado de cada elemento de cada columna:\n", cuadrado_columnas)
```
