#### 1. Introducción a Numpy 

```python
import numpy as np

# Creando un array a partir de una lista de Python
lista = [1, 2, 3, 4, 5]
# Creando un array de NumPy (ndarray) a partir de la lista
array = np.array(lista)

print(f"Array de Python: {lista} de tipo {type(lista)}")
print(f"Array de NumPy: {array} de tipo {type(array)}")
```

```python
# Creando un array a partir de una lista de Python
lista = [1, 2, 3, 4, 5]
# Creando un array de NumPy (ndarray) a partir de la lista
np_nros = np.array(lista)

## Otras propiedades de los arrays de NumPy
print("np_nros.dtype :", np_nros.dtype)
print("np_nros.ndim :", np_nros.ndim)
print("np_nros.size :", np_nros.size)
print("np_nros.shape :", np_nros.shape)
print("np_nros.itemsize :", np_nros.itemsize)
print("np_nros.data :", np_nros.data)
```
